```python
import pandas as pd
import numpy as np

from pathlib import Path
import sys
```

```python
PROCESSED_DATA_FOLDER = Path("../Data/Processed/").resolve()
```

```python
bicliques = pd.read_csv(PROCESSED_DATA_FOLDER / "bicliques_wave_5.csv", index_col=0)
```

```python
bicliques['found_rows'] = bicliques['found_rows'].map(
    lambda s: np.fromstring(s.strip("[ ]").replace('\n', ''), sep=' ', dtype=int)
)
bicliques['found_cols'] = bicliques['found_cols'].map(
    lambda s: np.fromstring(s.strip("[ ]").replace('\n', ''), sep=' ', dtype=int)
)
```

```python
data = pd.read_parquet(PROCESSED_DATA_FOLDER / "base_values_wave_5.parquet")
```

```python
complete_data = []
for _, bc in bicliques.iterrows():
    rows, cols = bc[["found_rows", "found_cols"]]
    complete_data.append(data.iloc[rows, cols])

assert all(df.notna().all().all() for df in complete_data)
```
